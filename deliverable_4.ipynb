## Task 1: Temperature

The `Temperature` class (in `temperature.py`) supports Celsius, Fahrenheit and
Kelvin. All conversions and comparisons go through Kelvin. Units are
case-insensitive, `value` and `unit` are read-only properties, and a
`ValueError` is raised for invalid units or temperatures below absolute zero.
Comparisons use a tolerance of 1e-9 and return `NotImplemented` for other types.

```python
from temperature import Temperature

t = Temperature(25, "c")
print(repr(t), "|", str(t))
print(t.to_fahrenheit(), t.to_kelvin())
print(t.convert_to("K"))
print(Temperature(0, "C") == Temperature(32, "F"))
```
